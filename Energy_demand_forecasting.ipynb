```python
import requests
import pandas as pd

url = 'https://api.neso.energy/api/3/action/datastore_search'
resource_id = '18c69c42-f20d-46f0-84e9-e279045befc6'
params = {'resource_id': resource_id, 'limit': 5000}

response = requests.get(url, params=params)
df = pd.DataFrame(response.json()['result']['records'])

# Convert Settlement_Date + Settlement_Period to datetime
df['datetime'] = pd.to_datetime(df['SETTLEMENT_DATE']) + pd.to_timedelta((df['SETTLEMENT_PERIOD'] - 1) * 30, unit='minutes')
df = df.set_index('datetime').sort_index()
df = df[['TSD', 'ND']]  # Use TSD (Transmission System Demand)

print(df.head())
```
